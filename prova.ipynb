```python
import pandas as pd
import numpy as np
from datasets import load_dataset
import json
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')
```

```python

def load_sportvu_dataset():
    """
    Load the SportVU dataset from HuggingFace
    Dataset: dcayton/nba_tracking_data_15_16

    Returns:
    - dataset: loaded dataset object
    """
    print("Loading SportVU dataset from HuggingFace...")
    print("Dataset: dcayton/nba_tracking_data_15_16")

    try:
        dataset = load_dataset(
            "dcayton/nba_tracking_data_15_16",
            "medium",
            split="train",
            trust_remote_code=True  # <-- Modifica per visualizzare dataset
        )
        print(f"Dataset loaded successfully: {len(dataset)} records available")

        # Display structure of first record
        print("\nFirst record keys:")
        print(dataset[0].keys())

        return dataset
    except Exception as e:
        print(f"Error loading dataset: {e}")
        return None

# Load the dataset
dataset = load_sportvu_dataset()
```

```python
def estrai_informazioni_partita(dati_partite):
    """
    Estrae gameid, gamedate, nome della squadra visitatrice e nome della squadra
    di casa da una lista di dizionari di dati di partite.

    Args:
        dati_partite (list): Una lista di dizionari, dove ogni dizionario
                            contiene i dati di una singola partita.

    Returns:
        list: Una lista di dizionari, ciascuno contenente:
              'gameid', 'gamedate', 'visitor_name', 'home_name'.
    """
    risultati = []

    for partita in dati_partite:
        # Estrazione delle chiavi di livello superiore
        game_id = partita.get('gameid')
        game_date = partita.get('gamedate')

        # Estrazione del nome della squadra visitatrice
        visitor_data = partita.get('visitor', {})
        visitor_name = visitor_data.get('name')

        # Estrazione del nome della squadra di casa
        home_data = partita.get('home', {})
        home_name = home_data.get('name')

        # Aggiunta dei dati estratti alla lista dei risultati
        risultati.append({
            'gameid': game_id,
            'gamedate': game_date,
            'visitor_name': visitor_name,
            'home_name': home_name
        })

    return risultati

# --- ESEMPIO DI UTILIZZO ---


# Chiama la funzione con i dati di esempio
dati_estratti = estrai_informazioni_partita(dataset)

# Stampa i risultati
for partita in dati_estratti:
    print(partita)
```

```python
# Continua dal codice di caricamento
# dataset = load_sportvu_dataset() 

def salva_sportvu_dataset(dataset, percorso_salvataggio):
    """
    Salva l'oggetto HuggingFace Dataset su disco.

    Args:
    - dataset: L'oggetto HuggingFace Dataset da salvare.
    - percorso_salvataggio: La cartella locale dove salvare il dataset.
    """
    if dataset is None:
        print("Impossibile salvare: Il dataset non è stato caricato correttamente.")
        return

    print(f"\nSalvataggio del dataset in corso in: '{percorso_salvataggio}'...")
    
    try:
        # Usa il metodo integrato della libreria Datasets
        # Questo salverà i dati in un formato efficiente (es. Arrow/Parquet)
        dataset.save_to_disk(percorso_salvataggio)
        
        print("✅ Dataset salvato con successo!")
        print(f"Per ricaricarlo, usa: loaded_dataset = load_from_disk('{percorso_salvataggio}')")
    
    except Exception as e:
        print(f"❌ Errore durante il salvataggio del dataset: {e}")

# --- Esecuzione ---

# 1. Definisci il percorso di salvataggio
cartella_locale = "fbx" 

# 2. Chiama la funzione di salvataggio
salva_sportvu_dataset(dataset, cartella_locale)
```
